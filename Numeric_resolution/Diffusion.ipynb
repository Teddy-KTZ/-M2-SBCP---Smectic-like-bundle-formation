## Mise en équation du problème

On considère une équation de diffusion : 

$$\partial_t\rho = D\nabla^2\rho$$

Avec $\rho$ la densité (ou concentration) en composé. On commencera en 1D puis on augmentera de dimension à 2D. On prends comme conditions initiale $\rho(x=0, t=0)=1$ et $\rho(x\neq0, t=0)=0$ . 
Pour commencer, on utilise une méthode d'Euler.

```python
x = np.arange(0, 10, 0.1)
t = np.arange(0, 10, 0.1)
print(x)

rho = np.empty(len(x), len(t))

print(rho)
```
